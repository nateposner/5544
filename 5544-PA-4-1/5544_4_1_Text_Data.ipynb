```python
emails = pd.read_csv('https://raw.githubusercontent.com/kevindavisross/data301/refs/heads/main/data/enron_email.csv')
corpus = emails[['body']].copy()
corpus
```

3\. Use `CountVectorizer` to create the TF matrix, and then convert it to a pandas data frame with appropriate column names. You can just use the defaults in `CountVectorizer`. Hint: this should pretty much be a straight copy of code from the reading, but make sure you understand the process behind the code.


```python
# ENTER YOUR CODE HERE. FEEL FREE TO ADD CELLS
from sklearn.feature_extraction.text import CountVectorizer
emails = emails.dropna(subset=['body'])
corpus = emails['body']
```

```python
vec = CountVectorizer()
vec.fit(corpus)
tf_sparse = vec.transform(corpus)
```

```python
tf = pd.DataFrame(
    tf_sparse.todense(),
    columns=vec.get_feature_names_out(),
    index=corpus.index
)
tf
```

4\. Explain what a document frequency (DF) is, and the idea of an IDF. Why might we use TF-IDF rather than just TF? This question is asking about concepts rather than computations.


**TF (term frequency) measures how much a term appears within a specific document.**

**DF (document frequency) measures how much a term appears across different documents, whereas IDF is a weight that makes words that only appear in less documents be a higher weight.**

**Using TF-IDF is useful because it weighs words based on their inter-document frequency vs. intra-document frequency, which can help us understand the importance of a word (basically, weighting words like "the" less and weighting certain keywords with high frequency in some documents higher).**

5\. Use `TfidfVectorizer` to create the TF matrix, and then convert it to a pandas data frame with appropriate column names. You can just use the defaults in `TfidfVectorizer`. Hint: this should pretty much be a straight copy of code from the reading, but make sure you understand the process behind the code.

```python
# ENTER YOUR CODE HERE. FEEL FREE TO ADD CELLS
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vec_tfidf = TfidfVectorizer()
vec_tfidf.fit(corpus)
tfidf_sparse = vec_tfidf.transform(corpus)
```

```python
tfidf = pd.DataFrame(
    tfidf_sparse.todense(),
    columns=vec_tfidf.get_feature_names_out(),
    index=corpus.index
)
tfidf
```

```python
tenemails = emails[emails['spam'].notna()]
tf_tenemails = tf.loc[tenemails.index]
tfidf_tenemails = tfidf.loc[tenemails.index]
```

```python
contains_word = (keywords_tf > 0)
percent_by_group = contains_word.groupby(tenemails['spam']).mean() * 100
percent_by_group
```

**Length of spam vs not spam emails**


```python
email_length = tf_tenemails.sum(axis=1)
email_length.groupby(tenemails['spam']).describe()
```

**Most common words in spam vs not spam**

```python
word_totals = tf_tenemails.groupby(tenemails['spam']).sum()
top_spam_words = word_totals.loc[1.0].sort_values(ascending=False).head(10)
top_spam_words
```

```python
top_norm_words = word_totals.loc[0.0].sort_values(ascending=False).head(10)
top_norm_words
```

8\. Consider just one email with unknown spam status, with `loc` 1980. Compute the cosine distance between the TF representation of this email and the TF representation of each of the emails with known spam status.

```python
from sklearn.metrics import pairwise_distances
```

```python
unknown_tf = tf.loc[[1980]]
tf_distances = pairwise_distances(unknown_tf, tf_tenemails, metric="cosine")[0]
tf_distances = pd.Series(tf_distances, index=tf_tenemails.index)
tf_distances
```

```python
# ENTER YOUR CODE HERE. FEEL FREE TO ADD CELLS
tf_sorted = tf_distances.sort_values()
tf_sorted.head(10)
```

11\. Now proceed in a similar manner to "guess" the spam status of each of the unknown emails.

```python
# ENTER YOUR CODE HERE. FEEL FREE TO ADD CELLS
unknown_emails = emails[emails['spam'].isna()]
tfidf_unknown = tfidf.loc[unknown_emails.index]
```

```python
all_distances = pairwise_distances(tfidf_unknown, tfidf_tenemails, metric="cosine")
all_distances = pd.DataFrame(all_distances, index=unknown_emails.index, columns=tfidf_tenemails.index)
```

```python
def majority_vote(row):
    nearest_5 = row.sort_values().head(5).index
    return tenemails.loc[nearest_5, 'spam'].mean()
```

```python
vote_share = all_distances.apply(majority_vote, axis=1)
guesses = (vote_share > 0.5).astype(int)
guesses
```

````python
def pluralize_gift(gift):
  """
  Returns plural of a noun

  Parameters
  ----------
  gift: str
    A noun

  Return
  ------
  str
    Plural version
  """

  if gift.find("oo") != -1:
    gift = gift.replace("oo", "ee")
  elif gift[-1] == "y":
    gift = gift[:-1] + "ies"
  else:
    gift = gift + "s"

  return gift
````

There are three basic ways to handle “ten” versus 10.

Tip
The Day.in.Words variable isn’t quite what you want! You want 12 to say "twelve" not "twelfth". Consider using .map() and a dictionary to make a column of the words you need. You can make this column outside of your functions, and add it to the data frame.

````python
def make_phrase(num, num_word, item, verb, adjective, location):
  """
  makes the phrase for each day

  Parameters
  ----------
  num: int
    The day number from 1 to 12
  num_word: str
    The day number as a word, like "one"
  item: str
    Gift in singular form
  verb, adjective, location: str
    The other parts of the phrase

  Return
  ------
  str
    The full phrase together
  """

  verb = "" if pd.isna(verb) else verb
  adjective = "" if pd.isna(adjective) else adjective
  location = "" if pd.isna(location) else location

  if num > 1:
    item = pluralize_gift(item)

  vowel_first = item[0] in "aeiou"
  if num == 1 and vowel_first:
    num_word = "an"
  elif num == 1:
    num_word = "a"

  phrase = num_word + " " + adjective + " " + item + " " + verb + " " + location

  return phrase.replace("  ", " ").strip()
````

````python
def sing_day(dataset, num, phrase_col):
  """
  Sings one verse of the song.

  Parameters
  ----------
  data frame
    Gift data with columns
  num: int
    Which day to sing about
  phrase_col: str
    Column name with each day's phrase

  Return
  ------
  str
    The verse for that day
  """

  num_word = dataset.loc[dataset["Day"] == num, "Day.in.Words"].iloc[0]
  intro = "On the " + num_word + " day of Christmas, my true love sent to me:"

  phrases = dataset[dataset["Day"] < num + 1].sort_values("Day", ascending=False)[phrase_col].tolist()
  if num == 1:
    gifts = phrases[0] + "."
  else:
    first_gifts = phrases[:-1]
    last_gift = phrases[-1]
    gifts = ",\n".join(first_gifts) + ", and\n" + last_gift + "."

  return intro + "\n" + gifts
````

Complete the following steps to decode the message.  

1. Remove any spaces before or after each word.
2. Any time you see the word "ugh", with any number of h's, followed by a punctuation mark, delete this.
3. No word should be longer than 16 characters. Drop all extra characters beyond 13 off the end of each word.
4. Replace all instances of exactly 2 a's with exactly 2 e's.
5. Replace all z's with t's.
6. Every word that ends in b, change that to a y.  *Hint: look out for punctuation!*
7. Every word that starts with k, change that to a v.  *Hint: look out for capitalization!*
8. Use `.join()` to recombine all your words into a message.
9. Find the movie this quote is from.

```python
# \w regular expression shortcut to say “match any character that might be found in a word” - i.e., not punctuation or whitespace.

# \b: “boundary” between word and non-word, such as punctuation or whitespace

# \s : “space” matches a single whitespace character

# \d : “digit” matches any single number 0-9

# ^ : matches the start of a string

# $ : matches the end of a string

# . : matches any character at all, except a new line (\n)
```

```python
# 1. Remove any spaces before or after each word.
nospace = message.str.strip()
new = ' '.join(nospace)
new
```

```python
# 2. Any time you see the word "ugh", with any number of h's, followed by a punctuation mark, delete this.
```

```python
re.findall(r"ugh+[^\w\s]", new)
''.join(new)
```

```python
new2 = re.sub(r'ugh+[^\w\s]', '', new)
new2
```

```python
# 3. No word should be longer than 16 characters. Drop all extra characters beyond 13 off the end of each word.
# wordmatch = r"\b([a-zA-Z]{13})[a-zA-Z]{4,}\b"
new3 = re.sub(r"\b([a-zA-Z]{13})[a-zA-Z]{4,}\b", r"\1", new2)
new3
```

```python
# 4. Replace all instances of exactly 2 a's with exactly 2 e's.
new4 = re.sub(r'aa', 'ee', new3)
new4
```

```python
# 5. Replace all z's with t's.
new5 = re.sub(r'z', 't', new4)
new5
```

```python
# 6. Every word that ends in b, change that to a y. Hint: look out for punctuation!
new6 = re.sub(r'b\b', 'y', new5)
new6
# new_text = re.sub(r'b\b', 'y', original_text)
```

```python
# 7. Every word that starts with k, change that to a v. Hint: look out for capitalization!
new7 = re.sub(r'\bk', 'v', new6)
new8 = re.sub(r'\bK', 'V', new7)
```

```python
# 8. Use .join() to recombine all your words into a message.
new8
```

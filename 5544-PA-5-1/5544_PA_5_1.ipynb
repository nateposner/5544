2\. Among all shows that aired in the U.S. on Feburary 4, 2018, which non-voice actors appeared on more than one show? Note: some people are credited with multiple rows on the same show and thus appear as multiple rows in the data frame.

Hint: You will need to write a for loop to make multiple requests. Use the "show.id" from the data from part 1, and use the ["Shows" endpoint](http://www.tvmaze.com/api#show-cast) to get the cast of each show. Don't forget to stagger your requests, or you will be blocked by the website!

```python
# YOUR CODE HERE. ADD CELLS AS NEEDED
import time
```

```python
show_ids = data_feb["show.id"].unique()
```

```python
cast = []
for show_id in show_ids:
    response = requests.get(f"https://api.tvmaze.com/shows/{show_id}/cast")
    members = response.json()
    for m in members:
        m["show_id"] = int(show_id)
    cast.extend(members)
    time.sleep(0.5)

len(cast)
```

```python
df_cast_feb = pd.json_normalize(cast)
non_voice = df_cast_feb[~df_cast_feb["voice"]]
shows_per_actor = non_voice.groupby("person.name")["show_id"].nunique()
shows_per_actor[shows_per_actor > 1].sort_values(ascending=False)
```

4\. For the recipes containing avocado, compute the proportion of reviews that are positive. For the avocado recipes with over 500 reviews, which one has the highest proportion of positive reviews?


```python
# YOUR CODE HERE. ADD CELLS AS NEEDED
df_avo = pd.json_normalize(data_avo)
df_avo[["name", "user_ratings.count_positive", "user_ratings.count_negative"]].head()
```

```python
df_avo["total_reviews"] = df_avo["user_ratings.count_positive"] + df_avo["user_ratings.count_negative"]
df_avo["prop_positive"] = df_avo["user_ratings.count_positive"] / df_avo["total_reviews"]
df_avo[["name", "total_reviews", "prop_positive"]].head()
```

```python
more_than_500 = df_avo[df_avo["total_reviews"] > 500]
more_than_500[["name", "total_reviews", "prop_positive"]].sort_values("prop_positive", ascending=False).head()
```
